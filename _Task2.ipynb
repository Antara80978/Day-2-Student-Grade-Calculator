# 🎓 Student Grade Calculator

### AI & ML Track — VEDA TECHNOLOGY — Task 2

This project calculates grades for students based on their marks
using Python conditional statements.

### Objectives
- Practice if, elif, and else statements
- Validate student marks
- Assign grades based on predefined ranges
- Process multiple students

### Grade function

```python
def calculate_grade(marks):
    """
    Assign a grade based on the student's marks.
    """

    if marks < 0 or marks > 100:
        return "Invalid"

    elif marks >= 90:
        return "A"

    elif marks >= 80:
        return "B"

    elif marks >= 70:
        return "C"

    elif marks >= 60:
        return "D"

    else:
        return "F"
```

### Test different marks


```python
test_marks = [95, 87, 76, 64, 52, 101, -5]

for marks in test_marks:
    grade = calculate_grade(marks)
    print(f"Marks: {marks:>3}  →  Grade: {grade}")
```

### Process Multiple Students

```python
students = {
    "Aarav": 95,
    "Priya": 87,
    "Rahul": 76,
    "Sneha": 64,
    "Aditya": 52,
    "Meera": 91,
    "Rohan": 83,
    "Ananya": 72
}

print("STUDENT GRADE REPORT")
print("-" * 35)

for name, marks in students.items():
    grade = calculate_grade(marks)
    print(f"{name:<10} | Marks: {marks:>3} | Grade: {grade}")
```

### Summary

```python
grade_count = {}

for marks in students.values():
    grade = calculate_grade(marks)

    if grade in grade_count:
        grade_count[grade] += 1
    else:
        grade_count[grade] = 1

print("\nGRADE SUMMARY")
print("-" * 25)

for grade, count in grade_count.items():
    print(f"Grade {grade}: {count} student(s)")
```

The program also validates marks to ensure that values outside the
0–100 range are treated as invalid.
